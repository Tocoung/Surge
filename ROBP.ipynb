```python
np.sort(w)
e=0.01/n
W=sum(w)+c
l=[]
l.append((0,1))
l.append((c+1,0))
keys = [k for k, _ in l]
for i in range(n - 1, -1, -1):
    ll=[]
    keys = [k for k, _ in l]
    j=bisect.bisect_right(keys,0)
    p1=l[j-1][1]
    j=bisect.bisect_right(keys,w[i])
    p2=l[j-1][1]
    p=(p1+p2)/2
    ll.append((0,p))
    pp=p/(1+e)
    while(pp>0):
        left=0
        right=W
        found=None
        while(left<right):
            mid=(left+right)//2
            j=bisect.bisect_right(keys,mid)
            p1=l[j-1][1]
            j=bisect.bisect_right(keys,mid+w[i])
            p2=l[j-1][1]
            p=(p1+p2)/2
            if(p>pp):
                left=mid+1
            else:
                found=(mid,p)
                right=mid
            # print((left,right))
        if(found):
            ll.append(found)
            pp=found[1]/(1+e)
        else:
            break
        #print(pp)
        #print(found)
    l=ll
    #print(l)
ans=l[0][1]*2**n
ans
```

```python
def knapsack_dp(n, w, c):
    # dp[i] = number of ways to achieve total weight i
    dp = [0] * (c + 1)
    dp[0] = 1  # One way to achieve 0 weight: pick nothing

    for i in range(n):
        for j in range(c, w[i] - 1, -1):  # go in reverse to avoid overcounting
            dp[j] += dp[j - w[i]]

    return dp
dp = knapsack_dp(n, w, c)
print(f"Total subsets with weight ≤ {c}: {sum(dp)}")  
```
